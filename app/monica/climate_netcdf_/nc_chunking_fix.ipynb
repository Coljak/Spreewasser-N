```python
import xarray as xr
import numpy as np
import dask.array as da
import os
from datetime import datetime
from pathlib import Path
```

```python
nc_files = [nc for nc in os.listdir('./') if nc.endswith('.nc')]
```

```python
for filename in nc_files:
    vari = filename.split('_')[1]
    if vari == 'sfcwind':
        vari = 'sfcWind'
    encoding = {vari: {
    'chunksizes': (61, 145, 109), 'zlib': True, 'complevel': 4 }}
    outputpath = f'climate_netcdf_chunked/{filename}'
    ds = xr.open_dataset(filename, chunks=None)
    ds.to_netcdf(outputpath, encoding=encoding)
```
